# Secant Method $x_{n+1} = x_n - f(x_n) \frac{x_n - x_{n-1}}{f(x_n) - f(x_{n-1})}$

```python
def function(x):
    return (x*x*x-3*x+12)
    
```

```python
def secant_method(f, x0, x1, tol=1e-8, max_iter=100):
    """
    Find a root of f(x) = 0 using the Secant Method.
    
    Parameters:
        f       : function whose root we want to find
        x0, x1  : initial guesses (should be close to the root)
        tol     : tolerance for convergence (default 1e-8)
        max_iter: maximum number of iterations
    
    Returns:
        Approximate root if found, else None
    """
    if abs(f(x1) - f(x0)) < 1e-12:
        print("Warning: Initial points too close or function flat.")
        return None
    
    for i in range(max_iter):
        # Secant formula
        x2 = x1 - f(x1) * (x1 - x0) / (f(x1) - f(x0))
        
        print(f"Iteration {i+1}: x = {x2:.10f}, f(x) = {f(x2):.2e}")
        
        # Check convergence
        if abs(x2 - x1) < tol:
            return x2
        
        # Update points
        x0 = x1
        x1 = x2
    
    print(f"Warning: Method did not converge after {max_iter} iterations.")
    return x1  # Return best approximation
```

```python
secant_method(f=function(), x0=1,x1=2)
```
