# Handling missing numerical data

## 1. Univariate Imputation:
###     by applying statistical techniques on the same column where imputation has to be done.
## 2. Multivariate Imputation:
###     by using other columns to impute missing value on one column.

<br><br>

# **Univariate Imputation**

# Mean/Median Imputation

### Use mean for normal distribution and median for skewed data.

## Cons: Changes shape of distribution, Outliers, Changes in correlation

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
x = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
x.head()
```

```python
X_train, Y_train, X_test, Y_test = train_test_split(x,y,test_size=0.2, random_state=2)
```

```python
X_train.shape, Y_train.shape
```

```python
X_train.isnull().mean()
```

```python
mean_age = X_train['Age'].mean()
mean_fare = X_train['Fare'].mean()
median_age = X_train['Age'].median()
median_fare = X_train['Fare'].median()
```

```python
X_train['Age_median'] = X_train['Age'].fillna(median_age)
X_train['Age_mean'] = X_train['Age'].fillna(mean_age)

X_train['Fare_median'] = X_train['Fare'].fillna(median_fare)
X_train['Fare_mean'] = X_train['Age'].fillna(mean_fare)
```

```python
X_train.sample(5)
```

```python
print('Original Age variable variance: ', X_train['Age'].var())
print('Age Variance after median imputation: ', X_train['Age_median'].var())
print('Age Variance after mean imputation: ', X_train['Age_mean'].var())

print('Original Fare variable variance: ', X_train['Fare'].var())
print('Fare Variance after median imputation: ', X_train['Fare_median'].var())
print('Fare Variance after mean imputation: ', X_train['Fare_mean'].var())
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)
X_train['Age'].plot(kind = "kde", ax = ax)
X_train['Age_mean'].plot(kind = "kde", ax = ax, color = 'green')
X_train['Age_median'].plot(kind = "kde", ax = ax, color = 'red')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)
X_train['Fare'].plot(kind = "kde", ax = ax)
X_train['Fare_mean'].plot(kind = "kde", ax = ax, color = 'green')
X_train['Fare_median'].plot(kind = "kde", ax = ax, color = 'red')
```

```python
X_train.corr()
```

```python
X_train[['Age', 'Age_median', 'Age_mean']].boxplot()
```

```python
X_train[['Fare', 'Fare_median', 'Fare_mean']].boxplot()
```

# Using Sklearn

```python
X_train, X_test, Y_train, Y_test = train_test_split(x,y, test_size=0.2, random_state=2)
```

```python
imputer1 =  SimpleImputer(strategy = 'median')
imputer2 =  SimpleImputer(strategy = 'mean')
```

```python
trf = ColumnTransformer([
    ('imp1', imputer1, ['Age']),
    ('imp2', imputer2, ['Fare'])
], remainder= 'passthrough')
```

```python
trf.fit(X_train)
```

```python
trf.named_transformers_['imp1'].statistics_  #median value of age
```

```python
trf.named_transformers_['imp2'].statistics_  #mean value of age
```

```python
X_train = trf.transform(X_train)
X_test = trf.transform(X_test)
```

# **Arbitrary value imputation**

### mostly used in categorical data. 

Benifit: Easy to apply  
Disadv: Pdf distort, vairance changes, covariance also changes

## used when data is not missing random 

<br> </br>

# **End of distribution imputation**

### If the data is normally distributed: use **mean + 3σ** or **mean - 3σ**

### If the data is skewed : IQR Proximity rule is used: **Q1 - 1.5*IQR**  or **Q3 - 1.5*IQR**

--> easy to use, covariance changes, variance changes 

used when value missing is not random
