# Handling dates and time

```python
import numpy as np 
import pandas as pd 
```

```python
date = pd.read_csv('orders.csv')
time = pd.read_csv('messages.csv')
```

```python
date.head()
```

```python
time.head()
```

```python
date.info(), time.info()
```

```python
# Converting to datetime datatype 
date['date'] = pd.to_datetime(date['date'])
```

# 1.Extract year

```python
date['date_year'] = date['date'].dt.year
```

```python
date['date_year']
```

```python
date['date_month'] = date['date'].dt.month
```

```python
date['date_month']
```

```python
date['date_month_name'] = date['date'].dt.month_name()
```

```python
date['date_month_name']
```

# for the day of week 

```python
date['dow'] = date['date'].dt.dayofweek
date.head()
```

```python
date['date_dow_name'] = date['date'].dt.day_name()
date['date_dow_name']
```

## is weekend? 

```python
date['is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1, 0)
date.head()
```

## Extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week
date.drop(columns=['product_id', 'city_id', 'orders']).head()
```

```python
date['date_quarter'] = date['date'].dt.quarter
```

```python
date['date_quarter']
```

## Extract Semester

```python
date['date_sem'] = np.where(date['date_quarter'].isin([1,2]), 1, 2)
date['date_sem']
```

## Extract time elapsed between two dates

```python
from datetime import datetime as dt
today = dt.today()
today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

## months passed 

```python
months = np.round((today - date['date']) / np.timedelta64(30, 'D'), 0)
```

```python
months
```
