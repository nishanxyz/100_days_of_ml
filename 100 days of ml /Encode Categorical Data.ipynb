# **One hot encoding(OHE): for nominal categorical data**

###  You expand the dataset using the values of the nominal categorical datas as the columns to use it as the input data. The new columns made are called dummy variables.
### Now one of the columns is dropped to make the input columns(x) independent.
### This is called eliminating mulitcollinearity (when independent variables in a regression model are highly correlated with each other).

```python
df = pd.read_csv('cars.csv')
```

```python
df.head()
```

```python
df['fuel'].nunique(), df['owner'].nunique()
```
